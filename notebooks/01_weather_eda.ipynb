Temperature, humidity, precipitation and wind are ERA5 reference values. Historical rain probability is missing. Hourly and daily profiles are descriptive; examine sample counts and partial local days before interpreting them.

```python
import json
from pathlib import Path

import pandas as pd
from IPython.display import Image, display

from mosamiat.backfill import digest

project_root = next(
    path for path in (Path.cwd(), *Path.cwd().parents) if (path / "src/mosamiat").is_dir()
)
OUTPUT = project_root / "artifacts/eda-smoke"  # Change for another published EDA run.
manifest = json.loads((OUTPUT / "manifest.json").read_text(encoding="utf-8"))
assert manifest["status"] == "complete", "Run mosamiat-eda first"
assert all(digest(OUTPUT / name) == checksum for name, checksum in manifest["sha256"].items())
summary = json.loads((OUTPUT / "summary.json").read_text(encoding="utf-8"))
metrics = pd.read_csv(OUTPUT / "metrics.csv", index_col="metric")
print("Provider reference data; not station observations or operational backtest evidence.")
```

```python
weather = [
    "temperature_c",
    "humidity_pct",
    "precipitation_mm",
    "rain_probability",
    "wind_speed_mps",
    "wind_direction_deg",
]
display(metrics.loc[weather])
display(pd.Series(summary["rain"]))
display(pd.read_csv(OUTPUT / "daily_precipitation.csv"))
display(pd.Series(summary["wind"]))
```

```python
for name in ["weather_series.png", "wind_rose.png", "distributions.png"]:
    display(Image(filename=str(OUTPUT / name), width=1000))
```

```python
display(pd.read_csv(OUTPUT / "hourly_profile.csv"))
display(pd.read_csv(OUTPUT / "daily_profile.csv"))
for name in ["hourly_profile.png", "daily_profile.png"]:
    display(Image(filename=str(OUTPUT / name), width=1000))
```

Raw wind degrees are excluded from arithmetic means, IQR flags and linear correlations. Sine/cosine components and a non-calm wind rose preserve circular meaning. Positive precipitation is not automatically a validated rain-event label. Seasonal and extreme-event conclusions require longer representative history.
